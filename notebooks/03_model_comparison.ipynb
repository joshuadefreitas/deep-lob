```python
from pathlib import Path
import json

import pandas as pd
```

```python
results_root = Path("..") / "results"

experiments = [
    "baseline_a",
    "h5_L100",
    "h10_L50",
    "h10_L200",
    "h20_L100",
]

experiments
```

```python
rows = []

for name in experiments:
    summary_path = results_root / name / "summary.json"
    if not summary_path.exists():
        print(f"[Warning] No summary.json for {name}, skipping.")
        continue

    with summary_path.open("r") as f:
        s = json.load(f)

    cfg = s["config"]
    lob = cfg["lob"]
    cls = s.get("classification_metrics", {})
    bt = s.get("backtest_stats", {})

    rows.append({
        "experiment": name,
        "window_size": lob["window_size"],
        "horizon": lob["horizon"],
        "threshold": lob["threshold"],
        "accuracy": cls.get("accuracy"),
        "macro_f1": cls.get("macro_f1"),
        "total_pnl": bt.get("total_pnl"),
        "sharpe": bt.get("sharpe_ratio"),
        "max_drawdown": bt.get("max_drawdown"),
        "n_trades": bt.get("n_trades"),
    })

df = pd.DataFrame(rows)
df
```
